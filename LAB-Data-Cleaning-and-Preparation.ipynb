## Iteration 2: Handling Missing Data

- Identify columns with missing data and the percentage of missing data in each column.
- For numerical columns, fill missing values with the mean of the column. (E.g., if price has missing values, fill them with the mean price.)
- For categorical columns, fill missing values with the mode of the column. (E.g., if category has missing values, fill them with the most frequent category.)
- Verify that all missing data has been handled appropriately.

````python
missing_percentage = df.isna().mean() * 100
df['price'] = df['price'].fillna(df['price'].mean())
df['units_sold'] = df['units_sold'].fillna(df['units_sold'].mean())
df['category'] = df['category'].fillna(df['category'].mode()[0])


#print(df.isnull().sum())
#df.info()
print(df.head(12))
````

## Iteration 4: Parsing Dates

In this iteration, your task is to properly handle and format any columns in the dataset that contain date information. Working with dates is a common requirement when analyzing time-based data, as it allows for more powerful insights such as trends over time, seasonality, or other temporal patterns.

The column we will focus on in this iteration is `sales_date`, which records the date of each sale. You will convert this column into a pandas datetime format, which allows for easier manipulation of the date information and enables the extraction of specific components, such as the year, month, and day.

#### **Objective:**
- Convert any columns with date information (e.g., `sales_date`) into pandas' `datetime` format.
- Extract the **year**, **month**, and **day** from the date column and create new columns for each. This step is useful if you later want to group data by month, year, or day for analysis.
- Set the `sales_date` column as the **index** of the DataFrame. This makes it easier to perform time-based indexing, filtering, or resampling.

````python
df['sales_date']=pd.to_datetime(df['sales_date'],format ='%Y-%m-%d')

df['year'] = df['sales_date'].dt.year
df['month'] = df['sales_date'].dt.month
df['day'] = df['sales_date'].dt.day

df.set_index('sales_date', inplace=True)
````

## Iteration 5: Creating Dummy Variables and Preventing Redundancy
In this iteration, we will convert the "region" column into dummy variables and ensure that we avoid the dummy trap by dropping one of the dummy variables. The dummy trap occurs when there is perfect multicollinearity between the dummy variables, which can confuse machine learning algorithms and lead to inaccurate results.

**Objective**:
- Convert the "region" column to dummy variables. We will use the pandas get_dummies() function, but we will drop one of the dummy variables to avoid redundancy.

- Avoid the dummy trap by using the drop_first=True option in get_dummies(), which removes the first dummy variable.

**Steps to Solve**:
- Generate dummy variables: Convert the "region" column into dummy variables and drop the first one to avoid the dummy trap.

- Concatenate dummy variables to the DataFrame: Join the newly created dummy variables with the original DataFrame.


````python
dummy_vars = pd.get_dummies(df['region'], prefix='region', drop_first=True).astype(int)

df = pd.concat([df, dummy_vars], axis=1)
````


## Iteration 6: Scaling and Normalization Techniques

In this iteration, your task is to apply **scaling** and **normalization** techniques to the numerical data in the dataset. Scaling and normalization are essential preprocessing steps in machine learning and data analysis, especially when working with algorithms that are sensitive to the ranges or distributions of the data.

Many machine learning models (such as linear regression, neural networks, and clustering algorithms) perform better when the numerical data is scaled or normalized because it helps avoid biases caused by different ranges or magnitudes in the features. For example, if one feature is measured in large units (like price) and another in small units (like the number of units sold), it can distort the model's performance if they are not scaled properly.

#### **Steps to solve:**

1. **Select the numerical columns:**
   Identify the numerical columns in your dataset that require scaling or normalization. For instance, in this dataset, numerical columns might include `price` and `units_sold`.

2. **Apply scaling techniques:**
   You will use **Scikit-learn** to apply two common scaling techniques:
   - **MinMaxScaler:** Scales the data to a fixed range, usually [0, 1]. It is useful when you need to preserve the relationships between values while bringing them into a standard range.
   - **StandardScaler:** Standardizes the data by subtracting the mean and scaling to unit variance (i.e., mean 0 and standard deviation 1). This is useful when the data has a normal distribution or when the model assumes normally distributed features.

   After applying each scaling method, you should compare the results to understand how each technique transforms the data.

3. **Compare the scaling results:**
   Compare the results of both scaling methods by printing a few rows of the scaled data. Based on the results, decide which method is most appropriate for the dataset.

````python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

data_to_scale = df[['price', 'units_sold']].copy()
````

````python
# Hacemos Estandarizacion
scaler_z = StandardScaler()

data_standardized = pd.DataFrame(scaler_z.fit_transform(data_to_scale), 
                                columns=['price_z', 'units_sold_z'])

print(data_standardized.describe().round(2)) # imprimimos para visualizar la estandarizacion
````

````python
# Hacemos Normalización
scaler_minmax = MinMaxScaler()
data_normalized = pd.DataFrame(scaler_minmax.fit_transform(data_to_scale), 
                              columns=['price_min-max', 'units_sold_min-max'])

print(data_normalized.describe().round(2))  # imprimimos para visualizar la normalizacion
````

````python
# unimos la Estandarizacion y la Normalizacion para COMPARAR
# borramos el index, para que no de problemas
df_reset = df.reset_index(drop=True)
data_standardized = data_standardized.reset_index(drop=True)
data_normalized = data_normalized.reset_index(drop=True)
# aqui lo unimos
df_scaled = pd.concat([df_reset, data_standardized, data_normalized], axis=1)
# aqui vemos 'price'
print(df_scaled[['price','price_z','price_min-max']].head())
````

````python
# aqui vemos 'units_sold'
print(df_scaled[['units_sold','units_sold_z','units_sold_min-max']].head())
````

La estandarización (Z-score) suele ser mejor que la normalización (Min-Max) en este dataset por 3 razones clave, 
evidentes al comparar los resultados del concatenado:

1. Robustez ante outliers:

- en 'price': Valores como 51.0 (mínimo) y 299.36 (máximo) son extremos.
- Normalización: Comprime todo el rango entre [0, 1], pero un solo outlier (ej. 300) distorsionaría la escala (ej. 50 → ~0.16, 200 → ~0.66).
- Estandarización: Mantiene la dispersión relativa (price_z = (x - mean) / std). Un outlier afecta menos (ej. media ≈ 166, std ≈ 70).

2. Distribución de los datos:

- en 'units_sold': Rango original: 1 a 50.
- Normalización: Convierte 1 → 0 y 50 → 1, pero valores intermedios (ej. 25 → 0.49) pierden contexto.
- Estandarización: Valores cerca de la media (≈25) ≈ 0, altos/bajos ≈ ±1.5, más intuitivo para modelos.
    
3. Compatibilidad con algoritmos:

- Modelos basados en distancia (KNN, SVM, PCA):
- Estandarización: Funciona mejor porque preserva las distancias relativas.
- Normalización: Si un outlier escala a 1, el resto se agrupa cerca de 0, sesgando resultados.

La estandarización es mejor aquí porque maneja outliers, conserva la interpretabilidad y es más útil para modelos estadísticos.
